```python
from pxr import Usd
import omni.usd

G1_ROOT = "/World/G1"   # change if your G1 uses another root path


def inspect_g1(robot, root_path=G1_ROOT, search_term="head"):
    stage = omni.usd.get_context().get_stage()

    root = stage.GetPrimAtPath(root_path)
    if not root.IsValid():
        raise RuntimeError(f"No valid USD prim found at {root_path}")

    print("=" * 80)
    print("ARTICULATION LINKS")
    print("=" * 80)

    if len(robot.link_paths) != 1:
        print(f"Warning: expected one articulation, got {len(robot.link_paths)}")

    articulation_paths = set()

    for name, path in zip(
        robot.link_names,
        robot.link_paths[0],
        strict=True,
    ):
        articulation_paths.add(path)
        print(f"{name:<40} {path}")

    print()
    print(f"Total articulation links: {len(robot.link_names)}")

    print("\n" + "=" * 80)
    print("ALL USD PRIMS UNDER G1")
    print("=" * 80)

    usd_prims = []

    for prim in Usd.PrimRange(root):
        path = str(prim.GetPath())
        name = prim.GetName()
        type_name = prim.GetTypeName()

        usd_prims.append((name, path, type_name))

        print(
            f"{name:<40} "
            f"type={type_name:<20} "
            f"path={path}"
        )

    print()
    print(f"Total USD prims: {len(usd_prims)}")

    print("\n" + "=" * 80)
    print(f"USD PRIMS MATCHING '{search_term}'")
    print("=" * 80)

    matches = [
        (name, path, type_name)
        for name, path, type_name in usd_prims
        if search_term.lower() in name.lower()
        or search_term.lower() in path.lower()
    ]

    if not matches:
        print(f"No USD prims containing '{search_term}' found.")
    else:
        for name, path, type_name in matches:
            is_link = path in articulation_paths

            print(
                f"name={name}\n"
                f"  path:              {path}\n"
                f"  type:              {type_name}\n"
                f"  articulation link: {is_link}\n"
            )

    print("\n" + "=" * 80)
    print("USD PRIMS THAT ARE NOT ARTICULATION LINKS")
    print("=" * 80)

    for name, path, type_name in usd_prims:
        if path not in articulation_paths:
            print(
                f"{name:<40} "
                f"type={type_name:<20} "
                f"path={path}"
            )
```
